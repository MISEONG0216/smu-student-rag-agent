## 7. Text2SQL 함수 구현

**TODO: 시스템 프롬프트를 팀 데이터 도메인에 맞게 수정하세요**

````python
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage

llm = init_chat_model("gpt-5.4-mini")


def text_to_sql(question: str, db: SQLDatabase) -> str:
    """
    자연어 질문을 SQL로 변환
    """

    system_prompt = f"""
당신은 상명대학교 2026학년도 2학기 개설강좌 데이터를 조회하는
PostgreSQL 전문가입니다.

사용자의 자연어 질문을 SQL 쿼리로 변환하세요.

[테이블 관계]
- course_schedule: 기본 시간표 테이블
- course_restrictions: 수강제한 정보
  - course_schedule."학수번호" = course_restrictions."학수번호"
  - 필요 시 "분반"도 함께 비교

- major_recognition: 타학과 전공인정 정보
  - course_schedule."학수번호" = major_recognition."학수번호"

[JOIN 규칙]
- 수강제한, 개설학과, 주관학과 관련 질문은 course_restrictions와 JOIN하세요.
- course_restrictions와 JOIN할 때는 가능하면 "학수번호"와 "분반"을 모두 비교하세요.
- 타학과 전공인정, 인정학과, 인정이수구분 관련 질문은 major_recognition과 JOIN하세요.
- JOIN 시 문자열 비교에는 TRIM()을 사용하세요.
- 같은 결과가 중복될 수 있으면 SELECT DISTINCT를 사용하세요.

데이터베이스 스키마:
{db.table_info}

[프로젝트 핵심 테이블]
- course_schedule: 상명대학교 2026학년도 2학기 개설강좌 및 시간표 데이터

[course_schedule 주요 컬럼]
- "No": 원본 데이터 번호
- "학년": 수강 대상 학년
- "이수구분": 전공선택 등 이수구분
- "학수번호": 교과목 코드
- "교과목명": 과목 이름
- "학점": 과목 학점
- "이론시간": 이론 수업시간
- "실습시간": 실습 수업시간
- "교양영역": 교양 영역
- "강의시간(강의실)": 수업 요일, 교시 및 강의실
- "분반": 분반
- "담당교수": 담당 교수
- "비고": 수강제한 및 기타 안내사항

[SQL 생성 규칙]
- PostgreSQL 문법을 사용하세요.
- 상명대학교 강좌 관련 질문에는 course_schedule 테이블을 사용하세요.
- SELECT 쿼리만 생성하세요.
- INSERT, UPDATE, DELETE, DROP, ALTER, CREATE는 생성하지 마세요.
- 한글 컬럼명은 반드시 큰따옴표로 감싸세요.
- 문자열 검색에는 ILIKE와 %를 활용하세요.
- 개수를 질문하면 COUNT를 사용하세요.
- 평균을 질문하면 AVG를 사용하세요.
- 가장 많거나 적은 것을 질문하면 ORDER BY를 사용하세요.
- 결과 개수를 지정하지 않은 목록 질문은 LIMIT 20을 사용하세요.
- SQL 코드만 반환하세요.
- 설명은 반환하지 마세요.
- 코드 블록은 사용하지 마세요.
- SQL은 세미콜론(;)으로 끝내세요.
- 과목 목록을 요청하는 질문이면 "교과목명"을 반드시 SELECT에 포함하세요.
- 표 형태로 답변할 경우 Markdown 표의 각 컬럼을 반드시 | 기호로 명확히 구분하세요.
"""

    messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=question)
    ]

    response = llm.invoke(messages)
    sql = response.content.strip()

    # 코드 블록 제거
    if sql.startswith("```"):
        lines = sql.split("\n")
        sql = "\n".join(lines[1:-1]) if len(lines) > 2 else sql
        sql = sql.replace("sql", "").replace("```", "").strip()

    return sql


print("✓ Text2SQL 함수 준비 완료")
````

## 9. 완전한 Text2SQL 시스템 (SQL 실행 + 자연어 답변)

**TODO: 답변 생성 프롬프트를 팀 데이터에 맞게 수정하세요**

````python
def query_database(question: str, db: SQLDatabase) -> str:
    """
    자연어 질문 → SQL 생성 → 실행 → 자연어 답변
    """

    # 1. SQL 생성
    print(f"[1] SQL 생성 중...")
    sql = text_to_sql(question, db)
    print(f"    {sql}\n")

    # 2. SQL 실행
    print(f"[2] SQL 실행 중...")
    try:
        result = db.run(sql)
        print(f"    실행 완료\n")
    except Exception as e:
        return f"SQL 실행 오류: {e}"

    # 3. 자연어 답변 생성
    print(f"[3] 답변 생성 중...")

    system_prompt = """
    당신은 상명대학교 2026학년도 2학기 개설강좌 및 수강정보 안내 전문가입니다.

    사용자의 질문과 SQL 조회 결과를 바탕으로
    학생이 이해하기 쉬운 자연어로 답변하세요.

    [답변 규칙]
    - 반드시 SQL 조회 결과에 근거해서 답변하세요.
    - 조회 결과에 없는 내용은 추측하지 마세요.
    - 과목명, 학점, 담당교수, 강의시간, 강의실 등의 정보를 정확하게 전달하세요.
    - 개수를 묻는 질문에는 총 개수를 명확하게 알려주세요.
    - 여러 과목이 조회된 경우 보기 쉽게 정리해서 답변하세요.
    - 데이터가 없는 경우 해당 조건에 맞는 데이터가 없다고 안내하세요.
    - 학생이 이해하기 쉬운 한국어로 답변하세요.
    """

    messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"""
질문: {question}

실행한 SQL:
{sql}

쿼리 결과:
{result}

위 SQL 조회 결과만을 근거로 사용자의 질문에 답변해주세요.
""")
    ]

    response = llm.invoke(messages)

    return response.content


print("✓ 완전한 Text2SQL 시스템 준비 완료")
````

````python
from IPython.display import Markdown, display

# 상명대학교 개설강좌 데이터 질문
question = "1학년 3학점 과목 중 담당교수와 강의시간을 포함해서 10개 알려줘"

print(f"질문: {question}\n")
print("="*80 + "\n")

answer = query_database(question, db)

print("\n" + "="*80)
print("\n답변:")
display(Markdown(answer))
````
